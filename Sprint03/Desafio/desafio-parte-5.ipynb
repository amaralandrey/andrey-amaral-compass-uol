#### 5. Mostre quantos apps são classificados como Mature +17.

Para encontrar a quantidade de aplicativos Mature +17 foi utilizada uma contagem simples com o método sum(). Foram encontrados 499 aplicativos Mature +17.

```python
import pandas as pd

play_store_data = pd.read_csv('/home/andrey/andrey-amaral-compass-uol-1/Sprint03/Desafio/googleplaystore.csv')

contagem_mature_17 = (play_store_data['Content Rating'] == 'Mature 17+').sum()

print(f"Aplicativos classificados como Mature 17+: {contagem_mature_17}")
```
