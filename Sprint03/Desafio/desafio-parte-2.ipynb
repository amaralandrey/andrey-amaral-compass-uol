#### 2. Faça um gráfico de barras contendo os top 5 apps por número de instalação.

Para a segunda parte, comecei procurando os 5 app mais instalados e para isso usei o método nlargest(), contudo antes de usar o método foi preciso tratar os dados para remover caracteres não númericos e a inconsistencia de uma das linhas ter a string 'Free' na coluna, no que optei por substituir a string por 0. 

```python
import pandas as pd

play_store_data = pd.read_csv('/home/andrey/andrey-amaral-compass-uol-1/Sprint03/Desafio/googleplaystore.csv')

play_store_data['Installs'] = play_store_data['Installs'].replace('Free', '0')

play_store_data['Installs'] = play_store_data['Installs'].str.replace('+', '').str.replace(',', '').astype(int)

mais_instalados = play_store_data.nlargest(5, 'Installs')

print(mais_instalados)
```

Em seguida, foi feita a plotagem do gráfico pedido.

```python
import matplotlib.pyplot as mpl

mpl.figure(figsize=(10, 6))

mpl.bar(mais_instalados['App'], mais_instalados['Installs'], color = 'green', width = 0.5)

mpl.title('5 Aplicativos Mais Instalados')

mpl.ylabel('Número de Instalações')

mpl.xticks(rotation=50, ha='right')

mpl.tight_layout()

mpl.show()
```
