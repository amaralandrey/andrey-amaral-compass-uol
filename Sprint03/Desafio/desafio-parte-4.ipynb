#### 4. Mostre qual o app mais caro existente no dataset.

Para encontrar o app mais caro, foi preciso tratar os dados da coluna 'Price', pois as linhas tinham os preços precididos de um cifrão, que foram substituídos por um espaço vazio, e algumas linhas continham strings, que foram substituídas por 0. O app mais caro foi encontrado com o método idxmax() em conjunto com o método .loc[]

```python
import pandas as pd

play_store_data = pd.read_csv('/home/andrey/andrey-amaral-compass-uol-1/Sprint03/Desafio/googleplaystore.csv')

play_store_data['Price'] = play_store_data['Price'].replace('Everyone', '0')

play_store_data['Price'] = play_store_data['Price'].str.replace('$', '').astype(float)

app_mais_caro = play_store_data.loc[play_store_data['Price'].idxmax()]

print(f"O aplicativo mais caro é: {app_mais_caro['App']} (preço: ${app_mais_caro['Price']})")
```
