#### 1. Leia o arquivo csv googleplaystore.csv e remova as linhas duplicadas.

Para a primeira parte, comecei contando quantos linhas duplicadas haviam no data set, usando para isso os métodos duplicated() e sum(). Havia 483 linhas duplicadas.

```python
import pandas as pd

play_store_data = pd.read_csv('/home/andrey/andrey-amaral-compass-uol-1/Sprint03/Desafio/googleplaystore.csv')

linhas_duplicadas = play_store_data.duplicated().sum()

print(f"Total linhas duplicadas: {linhas_duplicadas}")
```

Após, utilizei o método drop_duplicates() para remover as linhas duplicadas, atribuindo o dataset tratado para o mesmo arquivo fonte. 

```python
play_store_data = play_store_data.drop_duplicates()

linhas_duplicadas = play_store_data.duplicated().sum()

print(f"Total linhas duplicadas: {linhas_duplicadas}")
```
