#### 3. Faça um gráfico de pizza mostrando as categorias de apps existentes no dataset de acordo com a frequência em que elas aparecem.

Inicialmente, procurei a frequencia das categorias usando o método value_counts() na coluna Category, ranqueando-as de forma decrescente.

```python
import pandas as pd

play_store_data = pd.read_csv('/home/andrey/andrey-amaral-compass-uol-1/Sprint03/Desafio/googleplaystore.csv')

categorias = play_store_data['Category'].value_counts()

print(categorias)
```

Em seguida fiz a plotagem do gráfico de pizza.

```python
import matplotlib.pyplot as mpl

mpl.figure(figsize = (16, 14))

mpl.pie(categorias, labels = categorias.index, autopct='%1.1f%%', startangle=140)

mpl.rcParams.update({'font.size': 18})

mpl.title('Categorias de Aplicativos\n')

mpl.axis('equal') 

mpl.show()
```
