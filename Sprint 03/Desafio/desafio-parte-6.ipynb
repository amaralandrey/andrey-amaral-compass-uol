#### 6. Mostre o top 10 apps por número de reviews bem como o respectivo número de reviews. Ordene a lista de forma decrescente por número de reviews.

Para encontrar os 10 apps com mais reviews foi preciso converter a coluna 'Reviews' para valores númericos usando o método pandas.to_numeric(), trantando as linhas que não tivessem números com errors='coerce'. Foram mantidas apenas uma ocorrência para cada aplicativo, o que foi feito com o método drop_duplicates. Os 10 apps com mais reviews foram encontrados com o método .nlargest. A ordenação foi feita com .sort_values.

```python
import pandas as pd

play_store_data = pd.read_csv('/home/andrey/andrey-amaral-compass-uol-1/Sprint03/Desafio/googleplaystore.csv')

play_store_data['Reviews'] = pd.to_numeric(play_store_data['Reviews'], errors='coerce')

play_store_data = play_store_data.drop_duplicates(subset='App', keep='first')

apps_reviews = play_store_data.nlargest(10, 'Reviews')[['App', 'Reviews']]

apps_reviews = apps_reviews.sort_values(by='Reviews', ascending=False)

print("10 aplicativos com mais Reviews\n", apps_reviews)
```
